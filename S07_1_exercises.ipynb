### Exercise 4 (Medium)
Build a simple neural network for text classification using embeddings.

```python
import torch
import torch.nn as nn

class TextClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_classes):
        super().__init__()
        # Define: embedding layer, linear layer
        self.embedding_layer = nn.Embedding(vocab_size, embed_dim)
        self.linear_layer = nn.Linear(embed_dim, num_classes)

    def forward(self, x):
        # Embed -> mean pooling -> classify
        # Embedding
        embedded = self.embedding_layer(x)
        # Pooling
        pooled = embedded.mean(dim = 1)
        # Classify
        classfication_layer = self.linear_layer(pooled)
        return classfication_layer
```

```python
# Test with dummy data
vocab_size = 20
embed_dim = 8
num_classes = 3

model = TextClassifier(vocab_size, embed_dim, num_classes)

dummy_input = torch.tensor([
    [1, 5, 2, 9, 3],
    [4, 4, 7, 1, 0]
])

output = model(dummy_input)
print("Model output:", output)
print("Output shape:", output.shape)
```
