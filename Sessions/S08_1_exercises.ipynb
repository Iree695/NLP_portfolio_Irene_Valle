### Exercise 3 (Medium)
Build an LSTM-based sentiment classifier.

```python
class LSTMClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim, num_classes):
        super().__init__()
        # Define: embedding, lstm, fc
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, num_classes)
    
    def forward(self, x):
        # Embed -> LSTM -> take last hidden -> classify
        embedded = self.embedding(x)
        lstm_out, (hidden, cell) = self.lstm(embedded)
        # Take last hidden state
        last_hidden = hidden[-1]  # Shape: (batch_size, hidden_dim)
        output = self.fc(last_hidden)
        return output

# Test with dummy data
dummy_input = torch.randint(0, 1000, (1, 5))  # batch_size=1, seq_len=5
model = LSTMClassifier(vocab_size=1000, embed_dim=50, hidden_dim=20, num_classes=2)
output = model(dummy_input)
print("Output shape:", output.shape)
```

### Exercise 4 (Medium)
Implement a character-level language model with LSTM.

```python
class CharLSTM(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim):
        super().__init__()
        # Embedding -> LSTM -> Linear (predict next char)
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, x, hidden=None):
        x = self.embedding(x)
        out, hidden = self.lstm(x, hidden)
        logits = self.fc(out[:, -1])   # predict next char
        return logits, hidden


# Train on a simple text and generate characters

```
