### Exercise 1 (Easy)
Convert texts to Bag-of-Words representation using sklearn.

```python
from sklearn.feature_extraction.text import CountVectorizer

texts = ["I love this movie", "This movie is terrible", "Great film!", "Waste of time"]

# Create BoW representation
vector = CountVectorizer()
bow_matrix = vector.fit_transform(texts)
print("Vocabulary:", vector.get_feature_names_out())
print("Bow representation: \n", bow_matrix.toarray())
```

### Exercise 2 (Easy)
Convert the same texts to TF-IDF representation.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

texts = ["I love this movie", "This movie is terrible", "Great film!", "Waste of time"]

# Create TF-IDF representation
vectors = TfidfVectorizer()
tfidf_matrix = vectors.fit_transform(texts)
print("Vocabulary:", vectors.get_feature_names_out())
print("TF-IDF representation: \n", tfidf_matrix.toarray())
```

### Exercise 3 (Medium)
Train a Naive Bayes classifier for sentiment analysis.

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split

texts = ["I love this movie", "Great film", "Excellent acting", "Best movie ever",
         "Terrible movie", "Waste of time", "Awful acting", "Worst film"]
labels = [1, 1, 1, 1, 0, 0, 0, 0]  # 1=positive, 0=negative

# Train Naive Bayes classifier
# 1º Convert the text to a vector:
vectorizer = CountVectorizer()
x = vectorizer.fit_transform(texts)

# 2º Train-test split:
x_train, x_test, y_train, y_test = train_test_split(
    x, labels, test_size = .25, random_state = 42
)

# 3º With Naive Bayes classifier
classf = MultinomialNB()
classf.fit(x_train, y_train)

# 4º Evaluation
accuracy = classf.score(x_test, y_test)
print("Accuracy:",accuracy)
```

### Exercise 4 (Medium)
Train a Logistic Regression classifier and compare with Naive Bayes.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split

texts = ["I love this movie", "Great film", "Excellent acting", "Best movie ever",
         "Terrible movie", "Waste of time", "Awful acting", "Worst film"]
labels = [1, 1, 1, 1, 0, 0, 0, 0]  # 1=positive, 0=negative

# Train Logistic Regression and compare accuracy with Naive Bayes
# 1º Convert the text to a vector:
vectorizer = CountVectorizer()
x = vectorizer.fit_transform(texts)

# 2º Train-test split:
x_train, x_test, y_train, y_test = train_test_split(
    x, labels, test_size = .25, random_state = 42
)

# 3º With Naive Bayes classifier:
nb_classf = MultinomialNB()
nb_classf.fit(x_train, y_train)

# 4º With Logistic Regression:
log_classf = LogisticRegression()
log_classf.fit(x_train, y_train)

# 5º Evaluation and comparison between both Naive Bayes and logistic regression:
nb_accuracy = nb_classf.score(x_test, y_test)
print("Accuracy Naive:", nb_accuracy)

log_accuracy = log_classf.score(x_test, y_test)
print("Accuracy Logistic :", nb_accuracy)
```
