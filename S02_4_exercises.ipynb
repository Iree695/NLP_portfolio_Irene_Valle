### Exercise 1.1: Counting Words

Given the sentence from the slides: *"They picnicked by the pool, then lay back on the grass and looked at the stars."*

Write Python code to:
1. Count words excluding punctuation
2. Count words including punctuation as separate tokens
3. Count unique word types (case-insensitive)

````python
import re
# 1. Count words excluding punctuation
sentence = "They picnicked by the pool, then lay back on the grass and looked at the stars."
words_no_punct = re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?", sentence)
count_no_punct = len(words_no_punct)

print("Words (excluding punctuation):", count_no_punct, words_no_punct)
````

````python
import re
import string
# 2. Count words including punctuation
sentence = "They picnicked by the pool, then lay back on the grass and looked at the stars."
tokens_with_punct = re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?|[{}]".format(re.escape(string.punctuation)), sentence)
count_with_punct = len(tokens_with_punct)

print("Tokens (including punctuation):", count_with_punct, tokens_with_punct)
````

````python
# 3. Count unique types (case-insensitive)  do not count the repeteated
sentence = "They picnicked by the pool, then lay back on the grass and looked at the stars."
unique_types = set(w.lower() for w in words_no_punct)
unique_count = len(unique_types)

print("Unique word types (case-insensitive):", unique_count, sorted(unique_types))
````

### Exercise 1.2: Handling Disfluencies

The slides show the utterance: *"I do uh main- mainly business data processing"*

Write code to:
1. Count all tokens including disfluencies
2. Remove filled pauses (uh, um) and fragments (words ending with -)
3. Count "clean" words

````python
utterance = "I do uh main- mainly business data processing"

# 1) Tokenizar (palabras, fragmentos con -, y puntuación)
tokens = re.findall(r"[A-Za-z]+-?|[^\w\s]", utterance)
all_token_count = len(tokens)

# 2) Quitar disfluencias DEL TEXTO (uh/um) y fragmentos (xxxx-)
filled_pauses = {"uh", "um"}

cleaned_tokens = []
for tok in tokens:
    # si es puntuación, la dejamos
    if re.fullmatch(r"[^\w\s]", tok):
        cleaned_tokens.append(tok)
        continue
    low = tok.lower()
    if low in filled_pauses:   # elimina "uh" / "um"
        continue
    if tok.endswith("-"):      # elimina fragmentos tipo "sh-"
        continue
    cleaned_tokens.append(tok)

# Reconstruir la frase con espaciado “bonito” (sin espacio antes de puntuación)
clean_text = " ".join(cleaned_tokens)
clean_text = re.sub(r"\s+([.,!?;:])", r"\1", clean_text)

# 3) Contar "clean" words (solo palabras, sin puntuación)
clean_words = re.findall(r"[A-Za-z]+", clean_text)

print("Original:", utterance)
print("1) All tokens:", all_token_count, tokens)
print("2) Clean text:", clean_text)
print("3) Clean word count:", len(clean_words), clean_words)
````

### Exercise 2.1: Basic Pattern Matching

Using the regex patterns from the slides, write code to:
1. Find all words starting with capital letters
2. Find all digits in a text
3. Find words that are NOT capitalized

````python
import re
text = "Chapter 1: Down the Rabbit Hole. Alice was 7 years old in 1865."

# 1) All words starting with capital letters
capitalized_words = re.findall(r"\b[A-Z][a-z]*\b", text)

# 2) All digits in the text (each digit as a separate match)
digits = re.findall(r"\d", text)

# 3) Words that are NOT capitalized (start with a lowercase letter)
not_capitalized = re.findall(r"\b[a-z][a-z]*\b", text)

print("1) Capitalized words:", capitalized_words)
print("2) Digits:", digits)
print("3) Not capitalized words:", not_capitalized)
````

### Exercise 2.3: Anchors and Word Boundaries

Write patterns using anchors (^, $, \b) to:
1. Match lines starting with a capital letter
2. Match lines ending with a period
3. Find the word "the" as a complete word (not in "other" or "there")

````python
lines = [
    "The quick brown fox",
    "jumped over the lazy dog.",
    "other animals watched.",
    "There was nothing else."
]
````

````python
# 1
p1 = r"^[A-Z]"
matching_lines = []
for ln in lines:
    if re.search(p1, ln):
        matching_lines.append(ln)

# 2
p2 = r"\.$"
end_period = []
for ln in lines:
    if re.search(p2, ln):
        end_period.append(ln)

# 3
p3 = r"\bthe\b"
contains_the = []
for ln in lines:
    if re.search(p3, ln, flags=re.IGNORECASE):
        contains_the.append(ln)

# Prints:
print("Start with capital:", matching_lines)
print("End with period:", end_period)
print("Word 'the':", contains_the)
````

### Exercise 2.4: Substitutions and Capture Groups

1. Convert US dates (mm/dd/yyyy) to EU format (dd-mm-yyyy)
2. Anonymize email addresses by replacing with [EMAIL]
3. Convert "I'm" contractions to "I am"

````python
import re
text_dates = "Meeting on 10/15/2011 and 03/22/2024"
text_emails = "Contact john@example.com or jane@test.org"
text_contractions = "I'm happy and I'm excited"
# 1. Date format conversion (exact example from slides)
text_dates = "The date is 10/15/2011"
converted = re.sub(r'(\d{2})/(\d{2})/(\d{4})', r'\2-\1-\3', text_dates)
print(f"1. US to EU date: '{text_dates}' -> '{converted}'")

# 2. Email anonymization
text_emails = "Contact john@example.com or jane@test.org"
anon = re.sub(r'\b[\w.-]+@[\w.-]+\.\w+\b', '[EMAIL]', text_emails)
print(f"\n2. Anonymized: '{anon}'")

# 3. Contraction expansion
text_contractions = "I'm happy and I'm excited"
expanded = re.sub(r"I'm", "I am", text_contractions)
print(f"\n3. Expanded: '{expanded}'")

# EXPLANATION (Slides reference):
# Capture groups use parentheses () to store matched values.
# In the replacement string:
# - \1 refers to first captured group (month)
# - \2 refers to second captured group (day)
# - \3 refers to third captured group (year)
# The slides show: re.sub(r"(\d{2})/(\d{2})/(\d{4})", r"\2-\1-\3", string)
````

### Exercise 2.5: ELIZA-style Pattern Matching

The slides show ELIZA, an early chatbot using regex.
Implement simple ELIZA rules for:
1. "I'm [emotion]" -> "WHY DO YOU THINK YOU ARE [emotion]"
2. "I need [something]" -> "WHAT WOULD IT MEAN IF YOU GOT [something]"
3. Any input with "always" -> "CAN YOU THINK OF A SPECIFIC EXAMPLE?"

````python
# YOUR CODE HERE
def simple_eliza(user_input):
    text = user_input.strip()
    # Implement ELIZA rules
    # 1. I'm [emotion]
    m = re.search(r"\bI[' ]?m\s+(.+)$", text, flags=re.IGNORECASE)
    if m:
        emotion = m.group(1).strip()
        return f"WHY DO YOU THINK YOU ARE {emotion.upper()}"
    # 2. I need [something]
    m = re.search(r"\bI\s+need\s+(.+)$", text, flags=re.IGNORECASE)
    if m:
        something = m.group(1).strip()
        return f"WHAT WOULD IT MEAN IF YOU GOT {something.upper()}"
    # 3. Any input with "always"
    m = re.search(r"\balalways\b", text, flags=re.IGNORECASE)
    if m:
        emotion = m.group(1).strip()
        return f"CAN YOU THINK OF A SPECIFIC EXAMPLE?"
````

````python
# Test
test_inputs = [
    "I'm depressed",
    "I need a vacation", 
    "They always ignore me"
]
for s in test_inputs:
    print("User:", s)
    print("ELIZA:", simple_eliza(s))
    print()
````
